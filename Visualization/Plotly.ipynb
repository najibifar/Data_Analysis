```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import os
```

```python
src_file = os.path.join('data', 'raw', 'CO2 Emissions_Canada.csv')
image_dir = ('images')
```

```python
df = pd.read_csv('co2_Emissions_Canada.csv')
df.head()
```

```python
fig = px.histogram(df, x = 'Engine Size(L)')
fig.show()
```

```python
fig = px.histogram(
    df,
    x = "Fuel Consumption City (L/100 km)",
    color = "Cylinders",
    labels = {"Fuel Consumption City (L/100 km)":"annual fuel cost"},
    nbins=40,
    title = "co2_Emissions_Canada"
)
fig.show()
```

```python
fig = px.colors.qualitative.swatches()
fig.show()
```

```python
fig = px.histogram(
    df,
    x="Fuel Consumption City (L/100 km)",
    color="Cylinders",
    labels={"Fuel Consumption City (L/100 km)": "annual fuel cost"},
    nbins=40,
    title="co2_Emissions_Canada",
    color_discrete_sequence=px.colors.qualitative.Set2,
)
fig.show()
```

```python
fig = px.colors.sequential.swatches()
fig.show()
```

```python
fig = px.scatter(
    df.query("`CO2 Emissions(g/km)` > 0"),
    x="Fuel Consumption Comb (L/100 km)",
    y="CO2 Emissions(g/km)",
    hover_name="Model",
    size="Cylinders",
    color="Cylinders",
    hover_data=["Make", "CO2 Emissions(g/km)", "Cylinders"],
)
fig.show()

```

```python
fig = px.scatter_matrix(df, dimensions=["Cylinders", "CO2 Emissions(g/km)"], color="Make" )
fig.show()
```

```python
fig = px.sunburst(
    df.query("`CO2 Emissions(g/km)` > 200"),
    path=["CO2 Emissions(g/km)", "Cylindser"]
)
fig.show()
```

```python
fig = px.density_heatmap(df, x="CO2 Emissions(g/km)", y="Cylinders")
fig.show()
```
