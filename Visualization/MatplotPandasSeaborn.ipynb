```python
fig, ax = plt.subplots()
ax.boxplot(df['Combined Electricity Consumption'], label=['All'], showfliers=False)
ax.set(title='MPG Distribution', ylabel='MPG for vehicles');
```

```python
fig, (ax1, ax2) = plt.subplots(2)
ax1.hist(df['Combined Electricity Consumption'], range=(10, 50))
ax1.set(title='Vehicle MPG', xlabel='Combined MPG', ylabel='Number of cars')

ax2.boxplot(df['Combined Electricity Consumption'], labels=['All'], vert=False, showfliers=False)
ax2.set(ylabel='MPG for vehicles')

fig.suptitle('MPG Distribution', fontsize= 14, fontweight= 'bold')
```

```python
avg_by_year = df.groupby(['Year'], as_index=False).agg({'Engine Index':'mean'}).round(2)
avg_by_year
```
