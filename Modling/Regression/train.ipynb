```python
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
import numpy as np
import pandas as pd
import cv2
import os
import h5py
import pickle
```

```python
train_labels = os.listdir('train_data/')
train_labels
```

```python
train_labels.sort()
```

```python
X = []
y = []

for label in train_labels:
    dir_path = os.path.join('train_data', label)
    print('Current Folder:', label)

    for i in range(1, 81):
        file_path = os.path.join(dir_path, f"{i}.jpg")
        print('Current Image:', file_path)

        image = cv2.imread(file_path)
        if image is None:
            print('⚠ Not found or unreadable:', file_path)
            continue

        image = cv2.resize(image, (500, 500))

        # HSV histogram
        image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
        hist = cv2.calcHist(
            [image_hsv], [0, 1, 2], None,
            [8, 8, 8], [0, 256, 0, 256, 0, 256]
        )
        cv2.normalize(hist, hist)
        histogram_features = hist.flatten()

        # Hu Moments
        image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
        hu_features = cv2.HuMoments(cv2.moments(image_gray)).flatten()

        # Final feature vector
        features = np.hstack((histogram_features, hu_features))

        X.append(features)
        y.append(label)
```
