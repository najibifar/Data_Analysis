```python
from sklearn import datasets
import matplotlib.pyplot as plt
import math

data = datasets.load_digits()
n = len(data.images)

cols = 20
rows = math.ceil(n / cols)

plt.figure(figsize=(20, rows))

for i in range(n):
    plt.subplot(rows, cols, i + 1)
    plt.imshow(data.images[i], cmap='gray')
    plt.axis('off')

plt.show()
```
