```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

x1 = np.arange(-4, 1.25, 0.01)
x2 = np.arange(1.25, 4, 0.01)

styles = plt.style.available
n = len(styles)

cols = 4
rows = (n + cols - 1) // cols

fig, axes = plt.subplots(rows, cols, figsize=(16, 4*rows))
axes = axes.flatten()

for ax, style in zip(axes, styles):
    plt.style.use(style)
    ax.fill_between(x1, stats.norm.pdf(x1), facecolor='green')
    ax.fill_between(x2, stats.norm.pdf(x2), facecolor='blue')
    ax.set_title(style, fontsize=8)
    ax.grid(True)

for ax in axes[n:]:
    ax.axis('off')

plt.tight_layout()
plt.show()
```
