```python
df['fico'].plot(kind='box')
plt.show()
```

```python
df['int.rate'].plot(kind='box')
plt.show()
```

```python
df['SQRT_RATE'] = df['int.rate'] ** 0.5
```

```python
df['sqrt_rate'] = np.sqrt(df['int.rate'])
```

```python
df.head()
```

```python
print("Skewness {}".format(df['int.rate'].skew()))
print('Skewness of the SQRT {}'.format(df.SQRT_RATE.skew()))
print('')
print("kurtosis {}".format(df['int.rate'].kurt()))
print("kurtosis of the SQR {}".format(df.SQRT_RATE.kurt()))
```

```python
fig, axes = plt.subplots(1,2, figsize=(15, 5))
sns.displot(df['int.rate'], ax=axes[0])
sns.displot(df['sqrt_rate'], ax=axes[1])
plt.show()
```
