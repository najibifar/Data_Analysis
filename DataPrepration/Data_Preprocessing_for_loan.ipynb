```python
import os
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('loan_data.csv', index_col='credit.policy')
df.head()
```

```python
df['pub.rec'] = df['pub.rec'].astype('category')
df['not.fully.paid'] = df['not.fully.paid'].astype('category')
```

```python
df['fico'].plot(kind='box')
plt.show()
```

```python
df['int.rate'].plot(kind='box')
plt.show()
```

```python
df['SQRT_RATE'] = df['int.rate'] ** 0.5
```

```python
df['sqrt_rate'] = np.sqrt(df['int.rate'])
```

```python
df.head()
```

```python
print("Skewness {}".format(df['int.rate'].skew()))
print('Skewness of the SQRT {}'.format(df.SQRT_RATE.skew()))
print('')
print("kurtosis {}".format(df['int.rate'].kurt()))
print("kurtosis of the SQR {}".format(df.SQRT_RATE.kurt()))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.histplot(df['int.rate'], ax=axes[0])
sns.histplot(df['sqrt_rate'], ax=axes[1])

plt.show()
```

```python
import scipy.stats as stats
```

```python
df['ZR'] = stats.zscore(df['int.rate'])
```

```python
df.head()
```

```python
df[(df['ZR']<-3) | (df['ZR']>3)]
```

```python
df[(df['ZR']<-3) | (df['ZR']>3)].shape[0]
```

```python
df['LOG_RATE'] = np.log(df['int.rate'])

print("Skewness of LOG: ", df['LOG_RATE'].skew())
print("Kurtosis of LOG: ", df['LOG_RATE'].kurt())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 5))

sns.histplot(df['int.rate'], ax=axes[0])
sns.histplot(df['sqrt_rate'], ax=axes[1])
sns.histplot(df['LOG_RATE'], ax=axes[2])

plt.show()
```
